```python
import pandas as pd 
```

```python
data_retail = pd.read_csv('../Bai thi thu/Course Files/retail_sales_dataset.csv')
```

```python
data_retail['date'] = pd.to_datetime(data_retail['date'], dayfirst=True , errors ='coerce')


invalid_dates = data_retail['date'].isna().sum()
print(f" Số ngày không hợp lệ : {invalid_dates}")

print(f"Shape trước khi loại bỏ những ngày lỗi : {data_retail.shape}")
data_retail_subset = data_retail.dropna(subset = ['date'])

print(f"Shape sau khi loại bỏ những ngày lỗi : {data_retail_subset.shape}")

#Set index cho date 
data_retail_subset = data_retail_subset.set_index('date')

#Sort index 
data_retail_subset = data_retail_subset.sort_index()
print(data_retail_subset.head())

print("Phạm vi thời gian của dữ liệu :")
print(f" Bắt đầu từ ngày : {data_retail_subset.index.min()}")
print(f" Kết thúc từ ngày : {data_retail_subset.index.max()}")


print(f" Tổng số ngày trong dữ liệu : {(data_retail_subset.index.max() - data_retail_subset.index.min() ).days} ngày " )
```
