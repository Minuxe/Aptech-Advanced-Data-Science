## 1.1. Fancy Indexing trong NumPy

### 🎯 Kiến thức cốt lõi (20% → 80%)

**Fancy Indexing** = Dùng **mảng chỉ số (array of indices)** để truy cập hoặc sửa đổi nhiều phần tử cùng lúc.

NumPy có 3 cách truy cập phần tử:

| Cách | Ví dụ | Mô tả |
|------|-------|--------|
| Basic Indexing | `arr[2]` | Lấy 1 phần tử |
| Slicing | `arr[1:5]` | Lấy dải liên tiếp |
| **Fancy Indexing** | `arr[[0, 3, 7]]` | Lấy nhiều phần tử **bất kỳ** bằng mảng chỉ số |

### 🏪 Ví dụ thực tế: Quản lý kho hàng

Bạn có danh sách 10 sản phẩm, cần lấy thông tin sản phẩm số 1, 5, 8 (không liên tiếp):

```python
products = np.array(['Áo', 'Quần', 'Giày', 'Mũ', 'Túi', 'Kính', 'Đồng hồ', 'Vớ', 'Thắt lưng', 'Nhẫn'])
selected = products[[1, 5, 8]]  # Fancy indexing!
# Kết quả: ['Quần', 'Kính', 'Thắt lưng']
```

**Điểm mấu chốt:** Fancy indexing cho phép bạn "nhặt" phần tử ở bất kỳ vị trí nào, không cần liên tiếp — giống như bạn có danh sách mua sắm và chỉ lấy đúng những món cần.

````python
import numpy as np

# === FANCY INDEXING DEMO ===
prices = np.array([100, 250, 500, 150, 300, 80, 1200, 50, 400, 900])
print("Tất cả giá:", prices)

# 1. Fancy Indexing - Truy cập bằng mảng chỉ số
indices = np.array([0, 3, 7])  # Lấy sản phẩm vị trí 0, 3, 7
print("Fancy indexing [0,3,7]:", prices[indices])  # [100, 150, 50]

# 2. Fancy Indexing - Sửa đổi phần tử
prices[[0, 3, 7]] = 999  # Đặt giá mới cho 3 sản phẩm
print("Sau khi sửa:", prices)

# 3. So sánh với Boolean Indexing (tìm theo điều kiện)
prices2 = np.array([100, 250, 500, 150, 300])
print("Boolean indexing (giá > 200):", prices2[prices2 > 200])  # [250, 500, 300]

print("\n=> Fancy Indexing: Truy cập/sửa đổi bằng MẢNG CHỈ SỐ")
print("=> Boolean Indexing: Tìm theo ĐIỀU KIỆN (khác nhau!)")
````

## 1.2. GridSearchCV trong Scikit-learn

### 🎯 Kiến thức cốt lõi

**GridSearchCV** = Grid Search + Cross Validation

- **Grid Search**: Tạo "lưới" tất cả tổ hợp tham số → thử từng tổ hợp
- **Cross Validation**: Chia dữ liệu thành K fold, đánh giá trên từng fold → kết quả đáng tin cậy hơn

### 🍳 Ví dụ thực tế: Nấu phở hoàn hảo

Bạn muốn nấu phở ngon nhất, có 3 yếu tố cần thử:
- **Lửa**: nhỏ, vừa, lớn (3 mức)
- **Thời gian ninh**: 2h, 4h, 6h (3 mức)
- **Gia vị**: ít, vừa, nhiều (3 mức)

GridSearchCV = Thử TẤT CẢ 3×3×3 = 27 tổ hợp, mỗi tổ hợp nấu 5 lần (5-fold CV), rồi chọn tổ hợp ngon nhất!

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

param_grid = {
    'C': [0.1, 1, 10],           # 3 giá trị
    'kernel': ['rbf', 'linear'], # 2 giá trị
    'gamma': [0.01, 0.1, 1]      # 3 giá trị  
}
# Tổng: 3 × 2 × 3 = 18 tổ hợp, mỗi tổ hợp chạy 5-fold CV

grid = GridSearchCV(SVC(), param_grid, cv=5)
grid.fit(X_train, y_train)
print(grid.best_params_)  # Tổ hợp tốt nhất
```

**Ghi nhớ:** GridSearchCV = "Thử tất cả, chọn tốt nhất" + Cross Validation để đánh giá công bằng.

## 1.3. Learning Rate trong Gradient Descent

### 🎯 Kiến thức cốt lõi

**Gradient Descent** = Tìm điểm thấp nhất (minimum) của hàm mất mát bằng cách đi theo hướng gradient.

**Learning Rate (α)** = Kích thước bước đi mỗi lần cập nhật.

| Learning Rate | Kết quả |
|:---:|:---|
| Quá **nhỏ** | Hội tụ **chậm** (mất rất nhiều bước) |
| **Vừa phải** | Hội tụ **tốt** ✓ |
| Quá **cao** | **PHÂN KỲ** (divergence) — nhảy qua nhảy lại, không bao giờ tìm được minimum |

### 🏔️ Ví dụ thực tế: Xuống núi trong sương mù

Bạn đứng trên đỉnh núi, muốn xuống thung lũng (điểm thấp nhất), nhưng sương mù dày nên chỉ cảm nhận được độ dốc dưới chân:

- **Bước nhỏ (LR thấp):** An toàn nhưng mất cả ngày mới xuống được → *Hội tụ chậm*
- **Bước vừa (LR vừa):** Xuống nhanh, ổn định → *Hội tụ tốt* ✓
- **Bước quá lớn (LR cao):** Nhảy qua thung lũng, lên đỉnh bên kia, rồi lại nhảy ngược → *PHÂN KỲ — không bao giờ xuống được!* ✗

**Công thức cập nhật:**
$$\theta_{new} = \theta_{old} - \alpha \cdot \nabla J(\theta)$$

Nếu $\alpha$ quá lớn → $\theta$ nhảy quá xa → Loss tăng thay vì giảm → **PHÂN KỲ**

````python
import numpy as np
import matplotlib.pyplot as plt

# Minh họa Gradient Descent với Learning Rate khác nhau
def f(x): return x**2  # Hàm mất mát đơn giản
def grad(x): return 2*x  # Gradient

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
x_range = np.linspace(-5, 5, 100)

for ax, lr, title in zip(axes, [0.01, 0.3, 1.1], 
                          ['LR=0.01 (Quá nhỏ → Chậm)', 'LR=0.3 (Vừa → Tốt ✓)', 'LR=1.1 (Quá cao → PHÂN KỲ!)']):
    ax.plot(x_range, f(x_range), 'b-', linewidth=2)
    x = 4.0  # Điểm khởi đầu
    points = [x]
    for _ in range(15):
        x = x - lr * grad(x)
        points.append(x)
        if abs(x) > 100: break
    points = np.array(points)
    valid = np.abs(points) < 6
    ax.plot(points[valid], f(points[valid]), 'ro-', markersize=5)
    ax.set_title(title, fontsize=11, fontweight='bold')
    ax.set_ylim(-1, 30)
    ax.set_xlim(-6, 6)

plt.tight_layout()
plt.show()
print("=> Learning Rate quá CAO → PHÂN KỲ (divergence)")
````

## 1.4. Giả định "Naive" trong Naive Bayes

### 🎯 Kiến thức cốt lõi

**Naive Bayes** dựa trên định lý Bayes:

$$P(Y|X_1, X_2, ..., X_n) \propto P(Y) \cdot P(X_1|Y) \cdot P(X_2|Y) \cdot ... \cdot P(X_n|Y)$$

**Giả định "Naive" (ngây thơ): Các đặc trưng (features) ĐỘC LẬP CÓ ĐIỀU KIỆN với nhau, cho trước nhãn Y.**

Nghĩa là: Biết nhãn Y rồi, thì feature $X_1$ không ảnh hưởng đến $X_2$.

### 📧 Ví dụ thực tế: Bộ lọc Spam Email

Phân loại email là spam hay không:
- Feature 1: Có từ "miễn phí" không?
- Feature 2: Có từ "trúng thưởng" không?
- Feature 3: Có link lạ không?

**Giả định Naive:** Biết email là spam rồi → "miễn phí" và "trúng thưởng" xuất hiện **độc lập** với nhau.

Thực tế: 2 từ này thường đi cùng nhau trong spam → Giả định sai nhưng **mô hình vẫn hoạt động tốt** trong thực tế! Đó là lý do gọi là "ngây thơ" — giả định đơn giản nhưng hiệu quả bất ngờ.

**Tại sao giả định này quan trọng?** Không có nó, phải tính $P(X_1, X_2, ..., X_n | Y)$ — tổ hợp quá lớn. Với giả định, chỉ cần tính riêng từng $P(X_i|Y)$ — nhanh hơn rất nhiều!

### 🧠 Ghi nhớ
- "Naive" ≠ Kém hiệu quả
- "Naive" = Giả định **các features độc lập có điều kiện**
- Đây là giả định giúp đơn giản hóa tính toán

### 📋 TÓM TẮT NHÓM 1 + ĐÁP ÁN

| **2** | **b)** | GridSearchCV = Thử **tất cả tổ hợp** tham số trong lưới |
| **3** | **b)** | LR quá cao → **Phân kỳ** (divergence) |

## 2.1. Regularization: Ridge, Lasso, Elastic Net

### 🎯 Kiến thức cốt lõi

**Regularization** = Thêm "hình phạt" vào hàm mất mát để tránh overfitting.

| Kỹ thuật | Penalty | Công thức | Đặc điểm |
|:---:|:---:|:---:|:---|
| **Ridge** | L2 | $\lambda \sum \beta_j^2$ | Thu nhỏ hệ số → **KHÔNG** về 0 |
| **Lasso** | L1 | $\lambda \sum |\beta_j|$ | Ép hệ số về **ĐÚNG 0** → **Feature Selection!** |
| **Elastic Net** | L1 + L2 | $\lambda_1 \sum |\beta_j| + \lambda_2 \sum \beta_j^2$ | **Kết hợp cả hai** |

### 🏠 Ví dụ thực tế: Dự đoán giá nhà

Bạn có 100 đặc trưng: diện tích, số phòng, khoảng cách bệnh viện, màu sơn tường, ...

- **Ridge:** Thu nhỏ tất cả hệ số, giữ lại hết 100 features → Vẫn phức tạp
- **Lasso:** Ép hệ số "màu sơn tường" về 0 (không quan trọng) → Chỉ giữ 20 features quan trọng → **FEATURE SELECTION** ✓
- **Elastic Net:** Kết hợp — vừa chọn features (L1), vừa ổn định nhóm features tương quan (L2)

### 🧠 Ghi nhớ bằng hình ảnh
- **Ridge = Bóp nhẹ** (co lại nhưng không triệt hệ số)
- **Lasso = Cắt bỏ** (triệt hệ số về 0 → loại feature)
- **Elastic Net = Bóp + Cắt** (kết hợp cả hai)

````python
import numpy as np
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.datasets import make_regression

# Tạo dữ liệu với 20 features, chỉ 5 features thật sự quan trọng
X, y = make_regression(n_samples=100, n_features=20, n_informative=5, random_state=42)

# So sánh 3 kỹ thuật
ridge = Ridge(alpha=1.0).fit(X, y)
lasso = Lasso(alpha=1.0).fit(X, y)
enet = ElasticNet(alpha=1.0, l1_ratio=0.5).fit(X, y)

print("=== HỆ SỐ CỦA 20 FEATURES ===")
print(f"{'Feature':<10} {'Ridge':>10} {'Lasso':>10} {'ElasticNet':>12}")
print("-" * 45)
for i in range(20):
    r = ridge.coef_[i]
    l = lasso.coef_[i]
    e = enet.coef_[i]
    marker = " ← LASSO loại!" if abs(l) < 0.01 else ""
    print(f"  X_{i:<6} {r:>10.2f} {l:>10.2f} {e:>12.2f}{marker}")

print(f"\nSố features có hệ số ≠ 0:")
print(f"  Ridge:      {np.sum(np.abs(ridge.coef_) > 0.01)}/20 (giữ hết)")
print(f"  Lasso:      {np.sum(np.abs(lasso.coef_) > 0.01)}/20 (CHỌN LỌC!)")
print(f"  ElasticNet: {np.sum(np.abs(enet.coef_) > 0.01)}/20")
````

## 2.2. Cây quyết định (Decision Tree)

### 🎯 Kiến thức cốt lõi

Cấu trúc cây quyết định:

```
        [Thu nhập > 50tr?]     ← ROOT NODE (nút gốc) — Đại diện TOÀN BỘ dữ liệu
        /                \
   [Tuổi > 30?]      [Nợ > 0?]   ← INTERNAL NODE (nút trong)
   /         \        /       \
[Duyệt]  [Từ chối] [Từ chối] [Duyệt]  ← LEAF NODE (nút lá) — Quyết định cuối
```

| Thành phần | Vai trò |
|:---:|:---|
| **Root Node** | Nút trên cùng, đại diện **toàn bộ dữ liệu**, phân chia đầu tiên |
| Internal Node | Nút giữa, tiếp tục phân chia |
| Leaf Node | Nút cuối, đưa ra kết quả dự đoán |
| Branch | Đường nối giữa các nút |

### 🏦 Ví dụ thực tế: Ngân hàng duyệt vay

Root Node hỏi câu hỏi quan trọng nhất (ví dụ: Thu nhập > 50 triệu?) → chia toàn bộ hồ sơ thành 2 nhóm → mỗi nhóm lại được chia tiếp cho đến khi có quyết định cuối cùng (duyệt/từ chối).

## 2.3. Support Vector Machine (SVM)

### 🎯 Kiến thức cốt lõi

SVM tìm **siêu phẳng (hyperplane)** phân chia tốt nhất giữa 2 lớp, sao cho **margin** (khoảng cách) lớn nhất.

**Thuật ngữ quan trọng:**

| Thuật ngữ | Ý nghĩa |
|:---:|:---|
| **Hyperplane** | Mặt phẳng phân chia 2 lớp |
| **Margin** | Khoảng cách giữa hyperplane và điểm gần nhất |
| **Support Vectors** | Các điểm dữ liệu **GẦN hyperplane NHẤT** — quyết định vị trí hyperplane |

### 🚧 Ví dụ thực tế: Xây hàng rào giữa 2 đàn gà

Bạn có 2 đàn gà (đỏ và trắng) trong sân. Bạn cần xây hàng rào (hyperplane) chia 2 đàn:
- **Hàng rào** = Hyperplane
- **Khoảng trống 2 bên hàng rào** = Margin (càng rộng càng tốt)
- **Con gà đứng sát hàng rào nhất** = **SUPPORT VECTORS** (nếu con gà này di chuyển, hàng rào phải dịch theo!)

### 🧠 Ghi nhớ
Support Vector = "Người lính canh biên giới" — đứng gần ranh giới nhất, quyết định vị trí ranh giới.

### 📋 TÓM TẮT NHÓM 2 + ĐÁP ÁN

| Câu | Đáp án | Từ khóa ghi nhớ |
|:---:|:---:|:---|
| **5** | **b) Lasso** | L1 penalty ép hệ số về 0 → **Feature Selection** |
| **6** | **c)** | Elastic Net = **L1 + L2** kết hợp |
| **7** | **c) Root Node** | Nút gốc = toàn bộ dữ liệu, **phân chia đầu tiên** |
| **8** | **c) Support Vectors** | Điểm **gần hyperplane nhất** |

## 3.2. Kiểm định ADF (Augmented Dickey-Fuller)

### 🎯 Kiến thức cốt lõi

**Mục đích:** Kiểm tra chuỗi thời gian có **dừng (stationary)** hay không.

**Giả thuyết:**
- $H_0$: Chuỗi có **nghiệm đơn vị** → **CHƯA DỪNG** (non-stationary)
- $H_1$: Chuỗi **DỪNG** (stationary)

**Quy tắc:**

| p-value | Kết luận | Hành động |
|:---:|:---:|:---|
| p < 0.05 | Bác bỏ $H_0$ → **Chuỗi DỪNG** | Dùng trực tiếp |
| p ≥ 0.05 | Không bác bỏ $H_0$ → **Chuỗi CHƯA DỪNG** | **Cần sai phân** |

### 📈 Ví dụ trong đề: p-value = 0.35

0.35 >> 0.05 → Không bác bỏ $H_0$ → **Chuỗi CHƯA DỪNG** → Cần lấy sai phân (d = 1 trong ARIMA).

### 🌡️ Ví dụ thực tế: Nhiệt độ hàng ngày

- Nhiệt độ có xu hướng tăng (biến đổi khí hậu) → chuỗi chưa dừng → p-value lớn
- Sau khi lấy sai phân (chênh lệch nhiệt độ ngày này so với hôm qua) → chuỗi dừng → p-value nhỏ

### 🧠 Ghi nhớ
- **p nhỏ → dừng (stationary)** → tốt, dùng luôn
- **p lớn (như 0.35) → chưa dừng** → cần sai phân

## 3.3. Phương pháp Elbow trong K-Means

### 🎯 Kiến thức cốt lõi

**Vấn đề:** K-Means cần biết trước số cụm K → Làm sao chọn K tốt nhất?

**Phương pháp Elbow:**
1. Chạy K-Means với K = 1, 2, 3, ..., 10
2. Tính **WCSS** (Within-Cluster Sum of Squares = tổng khoảng cách trong cụm) cho mỗi K
3. Vẽ đồ thị K vs WCSS
4. Tìm điểm **"khuỷu tay" (elbow)** — nơi WCSS giảm chậm lại

```
WCSS
 |\                         
 | \                        
 |  \                       
 |   \_____  ← ELBOW (K tối ưu)
 |         \___________
 |_________________________ K
 1  2  3  4  5  6  7  8
```

### 🏪 Ví dụ thực tế: Mở chi nhánh cửa hàng

Bạn muốn mở các chi nhánh phục vụ khách hàng khắp thành phố:
- K=1: 1 cửa hàng → khách phải đi xa → WCSS rất lớn
- K=3: 3 cửa hàng → WCSS giảm mạnh (điểm elbow!)
- K=10: 10 cửa hàng → WCSS giảm thêm rất ít, nhưng chi phí tăng nhiều

→ K=3 là tối ưu (giảm WCSS nhiều nhất so với chi phí thêm)!

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# Tạo dữ liệu 3 cụm rõ ràng
X, _ = make_blobs(n_samples=300, centers=3, cluster_std=1.0, random_state=42)

# Chạy Elbow Method
wcss = []
K_range = range(1, 11)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X)
    wcss.append(km.inertia_)

# Vẽ đồ thị Elbow
plt.figure(figsize=(8, 4))
plt.plot(K_range, wcss, 'bo-', linewidth=2, markersize=8)
plt.axvline(x=3, color='r', linestyle='--', label='Elbow tại K=3')
plt.xlabel('Số cụm K')
plt.ylabel('WCSS (Within-Cluster Sum of Squares)')
plt.title('Phương pháp Elbow → Xác định SỐ CỤM TỐI ƯU')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("=> Elbow Method: Xác định SỐ CỤM TỐI ƯU cho K-Means")
````

## 4.1. DBSCAN (Density-Based Spatial Clustering)

### 🎯 Kiến thức cốt lõi

**DBSCAN** phân cụm dựa trên **MẬT ĐỘ điểm dữ liệu trong vùng lân cận** (ε-neighborhood).

**2 tham số quan trọng:**
- **ε (eps):** Bán kính vùng lân cận
- **MinPts:** Số điểm tối thiểu trong vùng ε để tạo cụm

**3 loại điểm:**

| Loại | Điều kiện | Ý nghĩa |
|:---:|:---:|:---|
| **Core Point** | ≥ MinPts điểm trong ε | Trung tâm cụm |
| **Border Point** | < MinPts nhưng gần Core Point | Biên cụm |
| **Noise Point** | Không thuộc cụm nào | Nhiễu |

### 🏘️ Ví dụ thực tế: Phân vùng dân cư

Dùng DBSCAN để tìm "khu dân cư" trên bản đồ:
- Đặt ε = 500m, MinPts = 5 (cần ít nhất 5 nhà trong bán kính 500m)
- Khu đông đúc → nhiều core points → 1 cụm
- Nhà riêng lẻ giữa cánh đồng → noise point

**So sánh với K-Means:**
- K-Means: Cần biết trước K, chỉ tìm cụm hình cầu
- DBSCAN: **TỰ TÌM** số cụm, phát hiện cụm hình dạng bất kỳ, loại nhiễu

## 4.2. Silhouette Score

### 🎯 Kiến thức cốt lõi

**Silhouette Score** đo chất lượng phân cụm cho từng điểm dữ liệu i:

$$s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}$$

- $a(i)$: Khoảng cách trung bình tới các điểm **cùng cụm** (càng nhỏ càng tốt)
- $b(i)$: Khoảng cách trung bình tới các điểm **cụm gần nhất khác** (càng lớn càng tốt)

**Giá trị từ -1 đến +1:**

| Score | Ý nghĩa |
|:---:|:---|
| **+1** | Hoàn hảo — nằm sâu trong cụm, xa cụm khác |
| **0** | Ở ranh giới giữa 2 cụm |
| **-1** | **Gán SAI cụm** — gần cụm khác hơn cụm hiện tại |

### 📌 Trong đề: Score = -0.3

Score **âm** → Điểm i gần cụm khác hơn cụm được gán → **CÓ THỂ BỊ GÁN SAI CỤM!**

### 🏫 Ví dụ thực tế: Xếp lớp học sinh

- Score = +0.8: Học sinh giỏi toán, xếp lớp toán → hợp lý!
- Score = 0: Học sinh giỏi cả toán lẫn văn → khó xếp
- Score = -0.3: Học sinh giỏi văn nhưng bị xếp nhầm lớp toán → **SAI CỤM!**

## 4.3. PCA và tầm quan trọng của Chuẩn hóa (Standardization)

### 🎯 Kiến thức cốt lõi

**PCA (Principal Component Analysis)** giảm chiều bằng cách tìm hướng có phương sai lớn nhất.

**Vấn đề:** PCA nhạy cảm với **scale** của dữ liệu.

Ví dụ:
- Feature 1: Lương (10,000 – 100,000) → Phương sai rất LỚN
- Feature 2: Tuổi (18 – 65) → Phương sai NHỎ

→ PCA sẽ chọn "Lương" làm thành phần chính chỉ vì nó có scale lớn hơn, **KHÔNG PHẢI** vì nó quan trọng hơn!

**Giải pháp: Chuẩn hóa (Standardization)** — đưa mỗi feature về **mean=0, std=1**

$$z = \frac{x - \mu}{\sigma}$$

→ Mọi feature đóng góp **ngang nhau**, PCA chọn đúng hướng quan trọng.

### 🏃 Ví dụ thực tế: Đánh giá vận động viên

So sánh VĐV bằng: Cân nặng (50-100 kg), Chiều cao (150-200 cm), Thời gian chạy (10-15 giây)
- Không chuẩn hóa: Chiều cao (150-200) sẽ áp đảo Thời gian (10-15) trong PCA
- Sau chuẩn hóa: Tất cả đều có mean=0, std=1 → đóng góp công bằng

### 🧠 Ghi nhớ
**PCA cần Standardization** vì PCA dựa trên phương sai → features scale lớn sẽ chiếm ưu thế nếu không chuẩn hóa.

## 4.4. t-SNE và Metric khoảng cách cho dữ liệu văn bản

## 5.1. Association Rules — Lift

### 🎯 Kiến thức cốt lõi

**3 chỉ số chính trong luật kết hợp:**

| Chỉ số | Công thức | Ý nghĩa |
|:---:|:---:|:---|
| **Support** | $P(A \cap B)$ | Tỉ lệ giao dịch chứa cả A và B |
| **Confidence** | $P(B|A) = \frac{P(A \cap B)}{P(A)}$ | Xác suất mua B khi đã mua A |
| **Lift** | $\frac{P(B|A)}{P(B)} = \frac{Confidence}{P(B)}$ | Mức độ **"tương tác"** giữa A và B |

**Giải mã Lift:**

| Lift | Ý nghĩa |
|:---:|:---|
| **Lift > 1** | A và B **tương quan dương** — mua A TĂNG xác suất mua B |
| **Lift = 1** | A và B **độc lập** — không liên quan |
| **Lift < 1** | A và B **tương quan ÂM** — mua A GIẢM xác suất mua B |

### 🛒 Ví dụ trong đề: Lift = 0.8

0.8 < 1 → **Tương quan ÂM** → Mua A hơi **GIẢM** xác suất mua B.

Ví dụ thực tế: {Kem đánh răng} ⇒ {Bàn chải} → Lift = 1.5 (tương quan dương, hay mua cùng)

{Sữa tươi} ⇒ {Sữa đậu nành} → Lift = 0.8 (tương quan âm — thường chọn 1 trong 2, không mua cả 2)

### 🧠 Ghi nhớ
- Lift < 1 = **Âm tính** (mua A giảm xác suất mua B)
- Lift > 1 = Dương tính (mua A tăng xác suất mua B)
- Lift = 0.8 ≠ 80% confidence!

## 5.2. FP-Growth — Thuật toán tìm tập phổ biến

### 🎯 Kiến thức cốt lõi

**FP-Growth** nhanh hơn Apriori vì KHÔNG tạo candidate sets. Hoạt động qua **2 PHA CHÍNH:**

| Pha | Thao tác |
|:---:|:---|
| **Pha 1** | Xây dựng **FP-Tree** (nén toàn bộ database vào cây) |
| **Pha 2** | Khai phá tập phổ biến từ FP-Tree (dùng conditional pattern base) |

### 🛒 Ví dụ thực tế: Phân tích giỏ hàng siêu thị

- **Pha 1 — Xây FP-Tree:** Quét tất cả hóa đơn, sắp xếp sản phẩm theo frequency giảm dần, nén vào cây
- **Pha 2 — Khai phá:** Từ cây, tìm ra các tổ hợp sản phẩm hay mua cùng (ví dụ: {Bia, Tã} → thường mua cùng)

### So sánh với Apriori
- **Apriori:** Quét database nhiều lần, tạo candidate sets → Chậm
- **FP-Growth:** Quét database 2 lần, nén vào cây → **Nhanh hơn nhiều**

## 5.3. Q-Learning — Off-policy

### 🎯 Kiến thức cốt lõi

**Công thức Q-Learning:**

$$Q(s, a) \leftarrow Q(s, a) + \alpha \Big[ r + \gamma \cdot \underset{a'}{\max} Q(s', a') - Q(s, a) \Big]$$

**Điểm mấu chốt:** Dùng $\max_{a'} Q(s', a')$ → Luôn chọn hành động **tốt nhất ở trạng thái tiếp theo** để cập nhật.

**On-policy vs Off-policy:**

| Loại | Hành vi | Ví dụ |
|:---:|:---:|:---|
| **On-policy** | Cập nhật dựa trên hành động **đã thực hiện** | SARSA |
| **Off-policy** | Cập nhật dựa trên hành động **tốt nhất có thể** (không nhất thiết đã thực hiện) | **Q-Learning** |

### 🎮 Ví dụ thực tế: Học chơi cờ

- **On-policy (SARSA):** Bạn ghi nhớ nước đi bạn **đã đi** → rút kinh nghiệm
- **Off-policy (Q-Learning):** Bạn ghi nhớ nước đi **TỐT NHẤT** có thể dù bạn đã đi nước khác → học từ nước đi tối ưu

Q-Learning có thể **khám phá** (explore) bằng ε-greedy nhưng **cập nhật** bằng hành động greedy → **OFF-POLICY**

## 5.4. Hệ số chiết khấu γ (Gamma/Discount Factor)

### 🎯 Kiến thức cốt lõi

$$G_t = r_t + \gamma \cdot r_{t+1} + \gamma^2 \cdot r_{t+2} + \gamma^3 \cdot r_{t+3} + ...$$

| γ | Hành vi |
|:---:|:---|
| **γ = 0** | $G_t = r_t$ → Chỉ quan tâm phần thưởng **TỨC THÌ**, bỏ qua tương lai |
| **γ = 0.5** | Cân bằng, phần thưởng xa giảm dần |
| **γ → 1** | Coi trọng phần thưởng tương lai gần như bằng hiện tại |

### 💰 Ví dụ thực tế: Quản lý tài chính

- **γ = 0:** Như người chỉ biết tiêu tiền hôm nay, không nghĩ đến ngày mai → Chỉ quan tâm **phần thưởng tức thì**
- **γ = 0.9:** Như người khôn ngoan, biết tiết kiệm hôm nay để hưởng lợi lâu dài
- **γ = 1:** Không phân biệt hiện tại và tương lai (hiếm khi dùng, có thể gây bất ổn)

### 🧠 Ghi nhớ
**γ = 0 → "Sống vội"** — chỉ thấy reward trước mắt, mù tương lai

## 6.1. Tính chất Markov trong MDP

### 🎯 Kiến thức cốt lõi

**MDP (Markov Decision Process)** là framework cho bài toán ra quyết định tuần tự.

**Tính chất Markov:** 
> **Trạng thái tương lai chỉ phụ thuộc vào trạng thái HIỆN TẠI và hành động được chọn — KHÔNG phụ thuộc lịch sử.**

$$P(s_{t+1} | s_t, a_t, s_{t-1}, a_{t-1}, ...) = P(s_{t+1} | s_t, a_t)$$

### 🎲 Ví dụ thực tế: Chơi cờ vua

- **Tính Markov:** Để quyết định nước đi tiếp theo, bạn chỉ cần nhìn **bàn cờ hiện tại** — không cần biết 20 nước đi trước đó.
- Mọi thông tin cần thiết đã được mã hóa trong trạng thái hiện tại (vị trí tất cả quân cờ).

### 🧠 Ghi nhớ
**Markov = "Quên quá khứ"** — Chỉ cần biết BÂY GIỜ ở đâu, không cần biết đi từ đâu đến.

## 6.2. Hàm kích hoạt ReLU — Chống Vanishing Gradient

### 🎯 Kiến thức cốt lõi

| Hàm | Công thức | Vấn đề |
|:---:|:---:|:---|
| **Sigmoid** | $\frac{1}{1+e^{-x}}$ | Gradient → 0 khi x lớn → **Vanishing gradient** |
| **Tanh** | $\frac{e^x - e^{-x}}{e^x + e^{-x}}$ | Tốt hơn Sigmoid, nhưng **vẫn vanishing** ở 2 đầu |
| **ReLU** ✓ | $\max(0, x)$ | Gradient = 1 khi x > 0 → **KHÔNG bị vanishing!** |
| **Softmax** | $\frac{e^{x_i}}{\sum e^{x_j}}$ | Dùng cho output layer (phân loại đa lớp) |

**Vanishing Gradient Problem:**
- Sigmoid/Tanh: Gradient nằm trong (0, 0.25) hoặc (0, 1)
- Qua nhiều layer: gradient nhân liên tục → *gần 0* → layer đầu không học được
- **ReLU:** Gradient = 1 (khi x > 0) → nhân bao nhiêu lần vẫn = 1 → **GIẢI QUYẾT vanishing!**

### 💡 Ví dụ thực tế: Truyền tin nhắn

Truyền tin nhắn qua 10 người (10 layers):
- **Sigmoid:** Mỗi người chỉ nhớ 25% → 10 người: $0.25^{10} ≈ 0.000001$ → tin nhắn mất gần hết!
- **ReLU:** Mỗi người nhớ 100% → 10 người: $1^{10} = 1$ → tin nhắn nguyên vẹn! ✓

### 🧠 Ghi nhớ
**ReLU = "Mặc định" cho hidden layers** vì chống Vanishing Gradient hiệu quả nhất.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 200)

# Các hàm kích hoạt
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)

# Gradient
sigmoid_grad = sigmoid * (1 - sigmoid)
tanh_grad = 1 - tanh**2
relu_grad = (x > 0).astype(float)

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
titles = ['Sigmoid', 'Tanh', 'ReLU ✓']
funcs = [sigmoid, tanh, relu]
grads = [sigmoid_grad, tanh_grad, relu_grad]
colors = ['red', 'blue', 'green']

for i, (title, func, grad, color) in enumerate(zip(titles, funcs, grads, colors)):
    axes[0][i].plot(x, func, color=color, linewidth=2)
    axes[0][i].set_title(f'{title} — Hàm', fontweight='bold')
    axes[0][i].grid(True, alpha=0.3)
    axes[0][i].axhline(y=0, color='k', linewidth=0.5)
    
    axes[1][i].plot(x, grad, color=color, linewidth=2)
    axes[1][i].set_title(f'{title} — Gradient', fontweight='bold')
    axes[1][i].grid(True, alpha=0.3)
    axes[1][i].axhline(y=0, color='k', linewidth=0.5)

plt.suptitle('So sánh hàm kích hoạt & Gradient\nReLU có gradient = 1 (x>0) → Không vanishing!', 
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 6.3. Backpropagation — Lan truyền ngược

### 🎯 Kiến thức cốt lõi

**Backpropagation** = Thuật toán tính gradient của hàm mất mát theo trọng số, rồi **điều chỉnh trọng số và bias** để giảm loss.

**Quy trình 2 bước:**
1. **Forward Pass:** Truyền input qua mạng → tính output → tính Loss
2. **Backward Pass (Backpropagation):** Truyền gradient từ output ngược lại → cập nhật trọng số

$$w_{new} = w_{old} - \alpha \cdot \frac{\partial Loss}{\partial w}$$

### 🎯 Ví dụ thực tế: Học bắn cung

1. **Forward:** Bắn mũi tên (forward pass) → thấy lệch phải 10cm (loss)
2. **Backward:** Phân tích nguyên nhân: góc tay, lực kéo, gió (gradient → mỗi yếu tố ảnh hưởng bao nhiêu)
3. **Update:** Chỉnh góc tay sang trái 5 độ, giảm lực kéo (cập nhật trọng số)
4. Lặp lại → mũi tên dần trúng đích (loss giảm)

### 🧠 Ghi nhớ
**Backpropagation = Tính gradient + Cập nhật trọng số/bias** → Mục đích: **GIẢM hàm mất mát**

## 6.4. GAN — Generator vs Discriminator

### 🎯 Kiến thức cốt lõi

**GAN (Generative Adversarial Network)** = 2 mạng đối kháng nhau:

| Thành phần | Nhiệm vụ | Ký hiệu |
|:---:|:---:|:---:|
| **Generator (G)** | Sinh dữ liệu **GIẢ** từ nhiễu ngẫu nhiên (z) → **đánh lừa** D | G(z) → fake data |
| **Discriminator (D)** | Phân biệt dữ liệu **THẬT** và **GIẢ** | D(x) → thật/giả |

```
Nhiễu z → [GENERATOR] → Ảnh giả → [DISCRIMINATOR] → Thật/Giả?
                                     ↑
                              Ảnh thật (từ dataset)
```

### 🎨 Ví dụ thực tế: Kẻ làm tiền giả vs Cảnh sát

- **Generator = Kẻ làm tiền giả:** Nhận "nguyên liệu" (noise z) → tạo tiền giả → cố gắng đánh lừa cảnh sát
- **Discriminator = Cảnh sát:** Nhận tiền (thật hoặc giả) → phân biệt thật/giả
- 2 bên cùng "tiến bộ": Kẻ làm giả ngày càng giỏi, cảnh sát ngày càng tinh → cuối cùng tiền giả y hệt thật!

### 🧠 Ghi nhớ
- **Generator: Sinh dữ liệu GIẢ từ NHIỄU** để đánh lừa Discriminator
- Discriminator: Phân biệt thật/giả

## 6.5. Scaled Dot-Product Attention — Chia cho √d_k

### 🎯 Kiến thức cốt lõi

**Công thức Attention:**

$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right) V$$

**Tại sao chia cho $\sqrt{d_k}$?**

Khi $d_k$ (chiều của key) lớn:
- Tích vô hướng $QK^T$ có giá trị **rất lớn**
- Đưa vào softmax → softmax **bão hòa** (output gần 0 hoặc 1)
- Gradient tại vùng bão hòa softmax → **gần 0** → không học được

Chia cho $\sqrt{d_k}$ → scale giá trị xuống → softmax hoạt động ở vùng gradient tốt.

### 🌡️ Ví dụ thực tế: Bình chọn nhà hàng

5 người cho điểm nhà hàng:
- **Không scale:** Điểm = [100, 200, 50, 300, 80] → softmax → [0.00, 0.00, 0.00, **1.00**, 0.00] → Chỉ chọn 1 nhà hàng, bỏ qua hết (bão hòa!)
- **Chia √d_k:** Điểm = [2, 4, 1, 6, 1.6] → softmax → [0.05, 0.15, 0.02, **0.70**, 0.08] → Vẫn ưu tiên nhà hàng tốt nhất, nhưng xem xét cả những nhà hàng khác

### 🧠 Ghi nhớ
**Chia √d_k → Ngăn tích vô hướng quá lớn → Softmax không bão hòa → Gradient không gần 0**

|:---:|:---:|:---|
| **21** | **b)** | Markov: Tương lai phụ thuộc **chỉ trạng thái hiện tại + hành động** |
| **22** | **c) ReLU** | Gradient = 1 → **Không vanishing** → Mặc định cho hidden layer |
| **23** | **a)** | Backpropagation: **Giảm loss** bằng điều chỉnh trọng số/bias |
| **24** | **c)** | Generator: **Sinh dữ liệu giả từ nhiễu** để đánh lừa Discriminator |
| **25** | **c)** | Chia √d_k → Ngăn **softmax bão hòa, gradient gần 0** |

## 7.1. Precision vs Recall — Hiểu đúng trade-off

### 🎯 Kiến thức cốt lõi

| Chỉ số | Công thức | Câu hỏi trả lời |
|:---:|:---:|:---|
| **Precision** | $\frac{TP}{TP + FP}$ | Trong các ca dự đoán Positive, bao nhiêu % **đúng thật**? |
| **Recall** | $\frac{TP}{TP + FN}$ | Trong tất cả ca Positive thật, mô hình **tìm được** bao nhiêu %? |

### 📌 Trong đề: Precision = 0.95, Recall = 0.40

- **Precision = 0.95:** Khi mô hình nói "Positive" → 95% là đúng → Đoán rất **chính xác**
- **Recall = 0.40:** Chỉ tìm được 40% ca Positive thật → **Bỏ sót 60%!**

→ Mô hình **chắc chắn khi dự đoán positive, nhưng bỏ sót đa số ca positive thật**.

### 🏥 Ví dụ thực tế: Chẩn đoán ung thư

- Precision = 0.95: Khi bác sĩ nói "Bạn bị ung thư" → 95% là đúng 
- Recall = 0.40: Chỉ phát hiện 40% bệnh nhân ung thư, **60% bị bỏ sót** → Nguy hiểm!

### 🧠 Ghi nhớ
- **Precision cao, Recall thấp** = "Khi nói thì đúng, nhưng nói ít" → Bỏ sót nhiều
- **Precision thấp, Recall cao** = "Nói nhiều, nhưng hay sai" → Báo nhầm nhiều

## 7.2. Đánh giá trên dữ liệu mất cân bằng (Imbalanced)

### 🎯 Kiến thức cốt lõi

**Vấn đề:** Tập dữ liệu 98% lớp 0, 2% lớp 1.

**Tại sao Accuracy không phù hợp?**
- Mô hình "ngu" luôn đoán lớp 0 → Accuracy = 98% → Có vẻ tốt nhưng thực tế **KHÔNG** phát hiện được bất kỳ ca lớp 1 nào!

**Chỉ số phù hợp:**

| Chỉ số | Tại sao phù hợp |
|:---:|:---|
| **F1-Score** | Trung bình điều hòa Precision & Recall → cân bằng cả hai |
| **AUC-PR** | Area Under Precision-Recall Curve → tập trung vào lớp thiểu số |
| ~~Accuracy~~ | ❌ Bị lớp đa số "đánh lừa" |
| ~~MSE, R²~~ | ❌ Dùng cho hồi quy, không phải phân loại |

### 💳 Ví dụ thực tế: Phát hiện giao dịch gian lận (fraud)

- 100,000 giao dịch: 98,000 hợp lệ, 2,000 gian lận
- Mô hình đoán hết "hợp lệ" → Accuracy = 98% → TƯỞNG TỐT!
- Nhưng F1-Score = 0 (không phát hiện gian lận nào)
- → Dùng **F1-Score hoặc AUC-PR** để đánh giá đúng

### 🧠 Ghi nhớ
**Dữ liệu mất cân bằng → KHÔNG dùng Accuracy → Dùng F1-Score hoặc AUC-PR**

## 7.3. Interoperability (Tính tương tác) trong ML

### 🎯 Kiến thức cốt lõi

**Interoperability** = Khả năng các hệ thống, mô hình, và công cụ **khác nhau** có thể **tích hợp và làm việc cùng nhau**.

| Khái niệm | Ý nghĩa |
|:---:|:---|
| **Interoperability** | Các hệ thống/mô hình/công cụ **tích hợp, làm việc cùng nhau** |
| Interpretability | Khả năng **giải thích** mô hình (khác nhau!) |

### 🏢 Ví dụ thực tế: Hệ sinh thái ML trong công ty

- Team A dùng TensorFlow, Team B dùng PyTorch
- Cả hai đều export model sang **ONNX format** → chạy trên cùng hệ thống production
- **MLflow** quản lý thí nghiệm cho cả hai framework
- → Đây là **Interoperability** — các hệ thống khác nhau có thể làm việc cùng nhau

Các công cụ hỗ trợ interoperability: ONNX, MLflow, Apache Arrow, PMML, ...

## 7.4. LIME — Giải thích mô hình cục bộ

### 🎯 Kiến thức cốt lõi

**LIME (Local Interpretable Model-agnostic Explanations):**
- **Local:** Giải thích từng dự đoán riêng lẻ (không phải toàn bộ model)
- **Interpretable:** Tạo mô hình đơn giản (linear) để giải thích
- **Model-agnostic:** Hoạt động với BẤT KỲ mô hình nào (black-box)

**LIME hỗ trợ:**

| Loại dữ liệu | Cách giải thích |
|:---:|:---|
| **Bảng (Tabular)** | Highlight features quan trọng nhất |
| **Văn bản (Text)** | Highlight từ/cụm từ quyết định |
| **Hình ảnh (Image)** | Highlight vùng ảnh (superpixels) quyết định |

→ **Hỗ trợ cả 3: Bảng, Văn bản, Hình ảnh!**

### 🏥 Ví dụ thực tế: Giải thích chẩn đoán

Mô hình dự đoán bệnh nhân A bị tiểu đường:
- LIME giải thích: "Vì glucose = 250 (rất cao) và BMI = 35 (béo phì)"
- Bác sĩ hiểu tại sao mô hình dự đoán vậy → tin tưởng hơn

### So sánh LIME vs SHAP
- **LIME:** Giải thích cục bộ (1 dự đoán), nhanh, đa dạng dữ liệu
- **SHAP:** Cả cục bộ lẫn toàn cục, dựa trên Shapley values (lý thuyết trò chơi)

## 7.5. Dynamic Computation Graph — PyTorch

### 🎯 Kiến thức cốt lõi

| Framework | Loại đồ thị | Đặc điểm |
|:---:|:---:|:---|
| **TensorFlow** (cũ) | Static Graph | Xây đồ thị trước → Chạy → Khó debug |
| **PyTorch** ✓ | **Dynamic Graph** | **Xây đồ thị khi chạy** → Linh hoạt, dễ debug |
| MLflow | Quản lý thí nghiệm | Không phải deep learning framework |
| Apache Arrow | Format dữ liệu | Không phải deep learning framework |

**Static vs Dynamic:**
- **Static (TF cũ):** Vẽ sơ đồ mạch điện → Hoàn thiện → Mới cắm điện chạy → Muốn thay đổi → vẽ lại!
- **Dynamic (PyTorch):** Vừa vẽ vừa chạy → Muốn thay đổi → thay ngay! → **Linh hoạt, dễ debug**

### 💻 Ví dụ code PyTorch

```python
import torch

# Dynamic: Đồ thị được tạo MỖI LẦN chạy forward
x = torch.tensor([1.0], requires_grad=True)
y = x * 2 + 3  # Đồ thị tính toán được tạo TẠI ĐÂY
y.backward()    # Gradient được tính ngược
print(x.grad)   # tensor([2.])
```

### 🧠 Ghi nhớ
**PyTorch = Dynamic Computation Graph** ← Từ khóa quan trọng nhất!

(Lưu ý: TensorFlow 2.x cũng hỗ trợ eager execution/dynamic graph, nhưng PyTorch là framework nổi tiếng nhất với dynamic graph ngay từ đầu)
