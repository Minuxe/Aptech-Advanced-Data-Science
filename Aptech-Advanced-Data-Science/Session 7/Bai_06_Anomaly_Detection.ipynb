## 1. Anomaly Detection Là Gì?

**Anomaly** (outlier/bất thường) = điểm dữ liệu **KHÁC BIỆT ĐÁNG KỂ** so với phần lớn.

### Ứng dụng thực tế:

| Lĩnh vực | Anomaly | Hậu quả nếu bỏ lỡ |
|:---|:---|:---|
| **Tài chính** | Giao dịch gian lận | Mất tiền khách hàng |
| **Y tế** | Chỉ số bất thường | Bỏ sót bệnh nhân nguy hiểm |
| **Sản xuất** | Sản phẩm lỗi | Giảm chất lượng |
| **Cybersecurity** | Truy cập bất thường | Bị hack |

### 3 Loại Anomaly

| Loại | Giải thích | Ví dụ |
|:---|:---|:---|
| **Point** | 1 điểm cá biệt bất thường | Giao dịch 100 triệu lúc 3h sáng |
| **Contextual** | Bất thường trong ngữ cảnh | 35 độ C bình thường mùa hè, bất thường mùa đông |
| **Collective** | Nhóm điểm bất thường cùng lúc | 100 giao dịch liên tiếp từ 1 thẻ |

## 2. Ba Thuật Toán Chính

### 2.1 Isolation Forest (QUAN TRỌNG NHẤT)

**Ý tưởng**: Anomaly dễ bị CÔ LẬP hơn điểm bình thường.

```
1. Chọn random 1 feature + random 1 ngưỡng cắt
2. Chia dữ liệu thành 2 phần (giống Decision Tree)
3. Lặp lại cho đến khi mỗi điểm bị cô lập

Anomaly: Cần ÍT lần cắt (path length ngắn)
Normal: Cần NHIỀU lần cắt (path length dài)
```

### 2.2 Local Outlier Factor (LOF)

**Ý tưởng**: So sánh MẬT ĐỘ ĐIỂM xung quanh.

```
LOF(x) = mật độ hàng xóm / mật độ x

LOF >> 1: x ở vùng THƯA hơn hàng xóm → Anomaly
LOF ~ 1: x có mật độ tương tự hàng xóm → Normal
```

### 2.3 One-Class SVM

```
Train trên dữ liệu BÌNH THƯỜNG
Tìm biên quyết định bao quanh dữ liệu normal
Điểm nằm NGOÀI biên → Anomaly
```

### So sánh:

| Thuật toán | Ưu điểm | Nhược điểm | Khi nào dùng |
|:---|:---|:---|:---|
| **Isolation Forest** | Nhanh, scale tốt | Cần tune contamination | Default choice |
| **LOF** | Phát hiện local anomaly | Chậm với data lớn | Mật độ không đều |
| **One-Class SVM** | Biên quyết định rõ | Chậm, nhạy kernel | Dữ liệu nhỏ-vừa |

````python
# Demo: So sanh 3 thuat toan tren du lieu 2D
np.random.seed(42)
X_normal = np.random.randn(300, 2) * 0.8
X_anomaly = np.random.uniform(-4, 4, (20, 2))
X_demo = np.vstack([X_normal, X_anomaly])
y_demo = np.array([1]*300 + [-1]*20)  # 1=normal, -1=anomaly

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

models = [
    ('Isolation Forest', IsolationForest(contamination=0.06, random_state=42)),
    ('LOF', LocalOutlierFactor(n_neighbors=20, contamination=0.06)),
    ('One-Class SVM', OneClassSVM(kernel='rbf', nu=0.06, gamma='auto'))
]

for ax, (name, model) in zip(axes, models):
    if name == 'LOF':
        preds = model.fit_predict(X_demo)
    else:
        preds = model.fit_predict(X_demo)
    
    normal_mask = preds == 1
    anomaly_mask = preds == -1
    
    ax.scatter(X_demo[normal_mask, 0], X_demo[normal_mask, 1], c='steelblue', s=20, alpha=0.5, label='Normal')
    ax.scatter(X_demo[anomaly_mask, 0], X_demo[anomaly_mask, 1], c='red', s=50, marker='x', label='Anomaly')
    
    n_detected = anomaly_mask.sum()
    ax.set_title(f'{name}\n(phat hien {n_detected} anomalies)', fontweight='bold')
    ax.legend()

plt.suptitle('So sanh 3 thuat toan Anomaly Detection', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 3. Tham Số Quan Trọng

### Isolation Forest:
- **contamination**: tỷ lệ anomaly dự kiến (0.01-0.1)
- **n_estimators**: số cây (default 100, thường đủ)
- **max_samples**: số mẫu mỗi cây (auto = min(256, n))

### LOF:
- **n_neighbors**: số hàng xóm (20 default, tăng nếu data lớn)
- **contamination**: tỷ lệ anomaly

````python
# Ảnh hưởng của contamination
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, cont in zip(axes, [0.02, 0.06, 0.15]):
    iso = IsolationForest(contamination=cont, random_state=42)
    preds = iso.fit_predict(X_demo)
    
    ax.scatter(X_demo[preds==1, 0], X_demo[preds==1, 1], c='steelblue', s=20, alpha=0.5)
    ax.scatter(X_demo[preds==-1, 0], X_demo[preds==-1, 1], c='red', s=50, marker='x')
    ax.set_title(f'contamination={cont}\n({(preds==-1).sum()} anomalies)', fontweight='bold')

plt.suptitle('Isolation Forest: Anh huong cua contamination', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print('contamination thap -> chi phat hien diem BAT THUONG NHAT')
print('contamination cao -> phat hien nhieu hon, co the BAO GOM NHIEU normal')
````

---

## 4. Anomaly Score

Ngoài label (+1/-1), có thể lấy **score liên tục** → xếp hạng mức độ bất thường.

````python
iso = IsolationForest(contamination=0.06, random_state=42)
iso.fit(X_demo)

# score_samples: THẤP = bất thường, CAO = bình thường
scores = iso.score_samples(X_demo)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sc = axes[0].scatter(X_demo[:,0], X_demo[:,1], c=scores, cmap='RdYlGn', s=30)
plt.colorbar(sc, ax=axes[0], label='Anomaly Score')
axes[0].set_title('Anomaly Score\n(DO = bat thuong, XANH = binh thuong)', fontweight='bold')

axes[1].hist(scores[y_demo==1], bins=30, alpha=0.7, label='Normal', color='steelblue')
axes[1].hist(scores[y_demo==-1], bins=15, alpha=0.7, label='Anomaly', color='red')
axes[1].axvline(x=iso.offset_, color='black', ls='--', label=f'Threshold={iso.offset_:.3f}')
axes[1].set_xlabel('Score'); axes[1].set_ylabel('Count')
axes[1].set_title('Score Distribution', fontweight='bold')
axes[1].legend()

plt.tight_layout()
plt.show()
print(f'Threshold (offset_): {iso.offset_:.3f}')
print('Score < threshold -> Anomaly')
````

## 5. Bài Thực Hành: Phát Hiện Gian Lận Thẻ Tín Dụng

````python
# Load data
df_fraud = pd.read_csv('../Bai thi thu/Course Files/credit_card_fraud_dataset.csv')
print(f'Shape: {df_fraud.shape}')
print(f'Columns: {df_fraud.columns.tolist()}')
print(df_fraud.head())
````

````python
# EDA
if 'IsFraud' in df_fraud.columns:
    target = 'IsFraud'
elif 'Class' in df_fraud.columns:
    target = 'Class'
else:
    target = df_fraud.columns[-1]

print(f'Target column: {target}')
print(f'\nPhan phoi:')
print(df_fraud[target].value_counts())
print(f'\nTy le fraud: {df_fraud[target].mean()*100:.2f}%')

fig, ax = plt.subplots(figsize=(6, 4))
df_fraud[target].value_counts().plot(kind='bar', ax=ax, color=['steelblue', 'red'])
ax.set_title(f'Phan phoi {target}\n(IMBALANCED!)', fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# Chuan bi features
numeric_cols = df_fraud.select_dtypes(include=[np.number]).columns.tolist()
if target in numeric_cols:
    numeric_cols.remove(target)
# Bo cot ID neu co
numeric_cols = [c for c in numeric_cols if 'id' not in c.lower()]

print(f'Features: {numeric_cols}')

X_fraud = df_fraud[numeric_cols].dropna()
y_fraud = df_fraud.loc[X_fraud.index, target]

scaler = StandardScaler()
X_fraud_scaled = scaler.fit_transform(X_fraud)
print(f'Shape: {X_fraud_scaled.shape}')
````

````python
# Isolation Forest
fraud_ratio = y_fraud.mean()
print(f'Ty le fraud thuc te: {fraud_ratio:.4f}')

iso_fraud = IsolationForest(contamination=min(fraud_ratio * 1.5, 0.5), 
                             random_state=42, n_estimators=200)
preds_iso = iso_fraud.fit_predict(X_fraud_scaled)
# Convert: -1 (anomaly) -> 1 (fraud), 1 (normal) -> 0
preds_binary = (preds_iso == -1).astype(int)

print(f'\nIsolation Forest Results:')
print(classification_report(y_fraud, preds_binary, target_names=['Normal', 'Fraud']))
````

````python
# LOF
lof = LocalOutlierFactor(n_neighbors=20, contamination=min(fraud_ratio * 1.5, 0.5))
preds_lof = lof.fit_predict(X_fraud_scaled)
preds_lof_binary = (preds_lof == -1).astype(int)

print('LOF Results:')
print(classification_report(y_fraud, preds_lof_binary, target_names=['Normal', 'Fraud']))
````

````python
# Visualization: PCA 2D
X_vis = PCA(2).fit_transform(X_fraud_scaled)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Nhan that
axes[0].scatter(X_vis[y_fraud==0, 0], X_vis[y_fraud==0, 1], c='steelblue', s=10, alpha=0.3, label='Normal')
axes[0].scatter(X_vis[y_fraud==1, 0], X_vis[y_fraud==1, 1], c='red', s=20, alpha=0.7, label='Fraud')
axes[0].set_title('Nhan that', fontweight='bold')
axes[0].legend()

# Isolation Forest
axes[1].scatter(X_vis[preds_binary==0, 0], X_vis[preds_binary==0, 1], c='steelblue', s=10, alpha=0.3, label='Normal')
axes[1].scatter(X_vis[preds_binary==1, 0], X_vis[preds_binary==1, 1], c='red', s=20, alpha=0.7, label='Fraud')
axes[1].set_title('Isolation Forest', fontweight='bold')
axes[1].legend()

# LOF
axes[2].scatter(X_vis[preds_lof_binary==0, 0], X_vis[preds_lof_binary==0, 1], c='steelblue', s=10, alpha=0.3, label='Normal')
axes[2].scatter(X_vis[preds_lof_binary==1, 0], X_vis[preds_lof_binary==1, 1], c='red', s=20, alpha=0.7, label='Fraud')
axes[2].set_title('LOF', fontweight='bold')
axes[2].legend()

plt.suptitle('Phat hien gian lan: Nhan that vs Mo hinh', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# Anomaly score ranking
scores_fraud = iso_fraud.score_samples(X_fraud_scaled)
df_result = df_fraud.loc[X_fraud.index].copy()
df_result['anomaly_score'] = scores_fraud
df_result['is_anomaly'] = preds_binary

print('Top 10 giao dich bat thuong nhat:')
top_anomalies = df_result.nsmallest(10, 'anomaly_score')
print(top_anomalies[[target, 'anomaly_score', 'is_anomaly'] + numeric_cols[:3]].to_string())

print(f'\nTrong top 10 bat thuong: {top_anomalies[target].sum()}/{len(top_anomalies)} la fraud that')
````

## 6. Đánh Giá Khi KHÔNG Có Nhãn

Trong thực tế, thường KHÔNG có nhãn anomaly. Cách đánh giá:

1. **Visual inspection**: PCA 2D, xem anomaly có ở vùng xa không
2. **Domain expert**: Cho chuyên gia review top anomalies
3. **Score distribution**: Bimodal = có anomaly rõ
4. **Consistency**: Nhiều thuật toán đều phát hiện cùng điểm = chắc chắn hơn

---

## TONG HOP: 20% Kien Thuc -> 80% Ung Dung

| # | Kiến thức cốt lõi | Chi tiết |
|:--|:---|:---|
| 1 | **Isolation Forest = default choice** | Nhanh, scale tốt, hoạt động trên mọi data |
| 2 | **contamination rất quan trọng** | Ước lượng tỷ lệ anomaly. Không biết -> bắt đầu 0.05 |
| 3 | **score_samples() cho ranking** | Không chỉ +1/-1, mà còn MỨC ĐỘ bất thường |
| 4 | **PCA 2D để visualize** | Luôn plot để kiểm tra kết quả |
| 5 | **Kết hợp nhiều thuật toán** | IF + LOF agree -> chắc chắn anomaly |

### Code Template:
```python
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

X_scaled = StandardScaler().fit_transform(X)

iso = IsolationForest(contamination=0.05, random_state=42)
labels = iso.fit_predict(X_scaled)  # 1=normal, -1=anomaly
scores = iso.score_samples(X_scaled)  # lower = more anomalous
```
