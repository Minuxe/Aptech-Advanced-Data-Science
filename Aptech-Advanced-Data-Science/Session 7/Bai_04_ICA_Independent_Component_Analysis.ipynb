## 1. ICA Là Gì?

### Ý tưởng cốt lõi

ICA tìm các **thành phần độc lập thống kê** (statistically independent) ẩn trong dữ liệu.

**Ví dụ kinh điển - Cocktail Party Problem:**
- 3 micro thu âm tại 1 bữa tiệc
- Mỗi micro ghi **hỗn hợp** 3 nguồn: tiếng người A, người B, nhạc nền
- ICA **tách ngược** ra 3 nguồn gốc từ 3 bản thu hỗn hợp

### Mô hình toán

```
X = A * S

X: dữ liệu quan sát (n_samples x n_features)
A: mixing matrix (ma trận trộn, KHÔNG biết)
S: nguồn gốc (sources, KHÔNG biết)

ICA tìm W = A^(-1) sao cho: S = W * X
```

### PCA vs ICA - Khác nhau CƠ BẢN

| Tiêu chí | PCA | ICA |
|:---|:---|:---|
| **Mục tiêu** | Max **variance** | Max **independence** |
| **Ràng buộc** | Components VUÔNG GÓC | Components ĐỘC LẬP THỐNG KÊ |
| **Gaussian** | Hoạt động tốt | CẦN non-Gaussian |
| **Thứ tự** | PC1 > PC2 > ... (theo variance) | KHÔNG có thứ tự |
| **Ứng dụng** | Giảm chiều | Tách nguồn (BSS) |

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA, FastICA
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
plt.rcParams['figure.dpi'] = 100
print('Thu vien san sang!')
````

## 2. Demo Cocktail Party: Tách Tín Hiệu Hỗn Hợp

````python
# Tạo 3 nguồn tín hiệu gốc
np.random.seed(42)
n = 2000
t = np.linspace(0, 8, n)

s1 = np.sin(2 * t)                    # Sóng sin
s2 = np.sign(np.sin(3 * t))           # Sóng vuông
s3 = np.random.laplace(size=n) * 0.5  # Nhiễu Laplace

S = np.c_[s1, s2, s3]  # 3 nguồn gốc

# Trộn (mixing) - giả lập 3 micro thu hỗn hợp
A = np.array([[1, 0.5, 0.3], [0.5, 1, 0.8], [0.3, 0.8, 1]])  # mixing matrix
X_mixed = S @ A.T  # 3 tín hiệu hỗn hợp

# Tách bằng ICA
ica = FastICA(n_components=3, random_state=42)
S_recovered = ica.fit_transform(X_mixed)

# So sánh với PCA
pca = PCA(n_components=3)
S_pca = pca.fit_transform(X_mixed)

fig, axes = plt.subplots(4, 3, figsize=(18, 14))

titles_row = ['Nguon GOC', 'Hon hop (Mixed)', 'ICA tach ra', 'PCA tach ra']
data_list = [S, X_mixed, S_recovered, S_pca]

for row, (data, row_title) in enumerate(zip(data_list, titles_row)):
    for col in range(3):
        axes[row, col].plot(t[:500], data[:500, col], lw=0.8)
        if col == 0:
            axes[row, col].set_ylabel(row_title, fontsize=11, fontweight='bold')
        axes[row, col].set_title(f'Signal {col+1}', fontsize=10)

plt.suptitle('ICA: Tach nguon tin hieu\nICA khoi phuc GAN DUNG nguon goc, PCA KHONG', 
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print('ICA tach duoc cac nguon doc lap (du thu tu va dau co the khac)')
print('PCA chi tim huong variance max, KHONG tach duoc nguon')
````

## 3. Khi Nào Dùng ICA?

| Ứng dụng | Chi tiết |
|:---|:---|
| **EEG/fMRI** | Tách tín hiệu não khỏi nhiễu |
| **Audio** | Tách giọng nói trong phòng ồn |
| **IoT Sensors** | Tách nguồn ảnh hưởng từ nhiều sensor |
| **Financial** | Tách yếu tố thị trường ẩn |

### Điều kiện để ICA hoạt động:
1. Sources phải **ĐỘC LẬP thống kê**
2. Sources phải **NON-GAUSSIAN** (ICA dựa vào non-Gaussianity)
3. Số sources <= số quan sát (n_components <= n_features)

## 4. Bài Thực Hành: ICA trên IoT Sensor Data

````python
# Load ICA dataset (IoT sensors)
df = pd.read_csv('../Bai thi thu/Course Files/ICA_dataset.csv')
print(f'Shape: {df.shape}')
print(f'Columns: {df.columns.tolist()}')
print(df.head())
````

````python
# Chọn features số
sensor_cols = ['Temperature', 'Humidity', 'Pressure', 'Co2 Gas', 'PM2.5', 'PM10']
available = [c for c in sensor_cols if c in df.columns]
print(f'Features: {available}')

X_sensor = df[available].dropna()
X_scaled = StandardScaler().fit_transform(X_sensor)
print(f'Shape sau drop NA: {X_scaled.shape}')

# Correlation truoc ICA
plt.figure(figsize=(8, 6))
sns.heatmap(pd.DataFrame(X_scaled, columns=available).corr(), annot=True, cmap='RdBu_r', center=0)
plt.title('Correlation TRUOC ICA\n(nhieu feature tuong quan)', fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# ICA
n_comp = min(4, len(available))
ica = FastICA(n_components=n_comp, random_state=42, max_iter=1000)
S_ica = ica.fit_transform(X_scaled)

# PCA de so sanh
pca = PCA(n_components=n_comp)
S_pca = pca.fit_transform(X_scaled)

fig, axes = plt.subplots(2, n_comp, figsize=(5*n_comp, 8))

for i in range(n_comp):
    axes[0, i].plot(S_ica[:500, i], lw=0.5)
    axes[0, i].set_title(f'ICA Component {i+1}', fontweight='bold')
    
    axes[1, i].plot(S_pca[:500, i], lw=0.5, color='orange')
    axes[1, i].set_title(f'PCA Component {i+1}', fontweight='bold')

axes[0, 0].set_ylabel('ICA', fontsize=12, fontweight='bold')
axes[1, 0].set_ylabel('PCA', fontsize=12, fontweight='bold')

plt.suptitle('ICA vs PCA: Tach thanh phan tu IoT Sensors', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# Kiem tra tinh doc lap: ICA components phai it tuong quan hon PCA
ica_corr = np.abs(np.corrcoef(S_ica.T))
pca_corr = np.abs(np.corrcoef(S_pca.T))

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.heatmap(ica_corr, annot=True, cmap='YlOrRd', ax=axes[0], fmt='.3f',
            xticklabels=[f'IC{i+1}' for i in range(n_comp)],
            yticklabels=[f'IC{i+1}' for i in range(n_comp)])
axes[0].set_title('ICA: Doc lap (off-diagonal ~ 0)', fontweight='bold')

sns.heatmap(pca_corr, annot=True, cmap='YlOrRd', ax=axes[1], fmt='.3f',
            xticklabels=[f'PC{i+1}' for i in range(n_comp)],
            yticklabels=[f'PC{i+1}' for i in range(n_comp)])
axes[1].set_title('PCA: Vuong goc (off-diagonal ~ 0)', fontweight='bold')

plt.tight_layout()
plt.show()

print('Ca ICA va PCA deu cho components it tuong quan')
print('Nhung ICA toi uu DOC LAP BAC CAO (higher-order independence)')
print('PCA chi toi uu VUONG GOC (uncorrelated, bac 2)')
````

````python
# Mixing matrix - feature nao dong gop vao IC nao
mixing = pd.DataFrame(ica.mixing_, index=available, 
                       columns=[f'IC{i+1}' for i in range(n_comp)])
plt.figure(figsize=(8, 5))
sns.heatmap(mixing, annot=True, cmap='RdBu_r', center=0, fmt='.2f')
plt.title('Mixing Matrix\n(feature nao tao thanh IC nao)', fontweight='bold')
plt.tight_layout()
plt.show()

print('Moi IC la 1 "nguon" an phia sau du lieu')
print('Mixing matrix cho biet moi feature la to hop cua cac nguon nhu the nao')
````

## TONG HOP: 20% Kien Thuc -> 80% Ung Dung

| # | Kiến thức cốt lõi | Chi tiết |
|:--|:---|:---|
| 1 | **ICA tách nguồn độc lập** | X = A*S, tìm W sao cho S = W*X |
| 2 | **PCA: max variance. ICA: max independence** | Mục tiêu khác nhau = ứng dụng khác nhau |
| 3 | **FastICA trong sklearn** | `FastICA(n_components=k).fit_transform(X)` |
| 4 | **Cần non-Gaussian** | ICA không tách được nếu sources là Gaussian |
| 5 | **Mixing matrix giải thích kết quả** | `ica.mixing_` cho biết feature đóng góp IC nào |

### Code Template:
```python
from sklearn.decomposition import FastICA
from sklearn.preprocessing import StandardScaler

X_scaled = StandardScaler().fit_transform(X)
ica = FastICA(n_components=k, random_state=42)
S = ica.fit_transform(X_scaled)       # Independent components
A = ica.mixing_                       # Mixing matrix
```
