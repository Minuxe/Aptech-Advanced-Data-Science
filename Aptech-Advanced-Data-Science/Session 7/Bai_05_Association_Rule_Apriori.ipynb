## 1. Association Rules Là Gì?

**Bài toán**: Tìm các mặt hàng thường được MUA CÙNG NHAU.

### Ví dụ thực tế:
- Khách mua **bia** thường mua kèm **snack** → Đặt gần nhau
- Khách mua **bỉm** thường mua kèm **sữa** → Bundle promotion
- Amazon: "Customers who bought X also bought Y"

### 3 Chỉ Số Quan Trọng

| Chỉ số | Công thức | Ý nghĩa |
|:---|:---|:---|
| **Support(A)** | P(A) = so don co A / tong don | A xuat hien thuong xuyen? |
| **Confidence(A->B)** | P(B given A) = Support(A,B) / Support(A) | Neu mua A, kha nang mua B? |
| **Lift(A->B)** | Confidence(A->B) / Support(B) | A va B lien quan THAT SU? |

### Diễn giải Lift:

| Lift | Ý nghĩa |
|:---|:---|
| **> 1** | A và B MUA CÙNG nhiều hơn ngẫu nhiên |
| **= 1** | Độc lập (không liên quan) |
| **< 1** | A và B ÍT MUA CÙNG hơn ngẫu nhiên |

### Thuật toán Apriori:

```
1. Tìm tất cả itemsets có support >= min_support
2. Từ frequent itemsets → sinh ra rules
3. Lọc rules theo confidence >= min_confidence

Nguyên lý Apriori: Nếu itemset X KHÔNG frequent,
thì mọi superset của X cũng KHÔNG frequent → cắt tỉa nhanh
```

## 2. Demo Đơn Giản: Hiểu Support, Confidence, Lift

````python
# Du lieu mau
transactions = [
    ['Bread', 'Milk', 'Eggs'],
    ['Bread', 'Butter'],
    ['Milk', 'Butter', 'Eggs'],
    ['Bread', 'Milk', 'Butter', 'Eggs'],
    ['Bread', 'Milk'],
    ['Milk', 'Eggs'],
    ['Bread', 'Eggs'],
    ['Bread', 'Milk', 'Eggs', 'Butter']
]

# Chuyen thanh one-hot encoding
te = TransactionEncoder()
te_data = te.fit_transform(transactions)
df_demo = pd.DataFrame(te_data, columns=te.columns_)
print('One-hot encoding:')
print(df_demo.astype(int).to_string())
print(f'\nSupport cua tung item:')
for col in df_demo.columns:
    print(f'  {col}: {df_demo[col].mean():.3f} ({df_demo[col].sum()}/{len(df_demo)} don)')
````

````python
# Apriori: Tim frequent itemsets
freq_items = apriori(df_demo, min_support=0.3, use_colnames=True)
freq_items = freq_items.sort_values('support', ascending=False)
print(f'So frequent itemsets: {len(freq_items)}')
print(freq_items.head(10).to_string())
````

````python
# Sinh rules
rules = association_rules(freq_items, metric='confidence', min_threshold=0.5)
rules = rules.sort_values('lift', ascending=False)

# Hien thi dep
display_cols = ['antecedents', 'consequents', 'support', 'confidence', 'lift']
rules_display = rules[display_cols].copy()
rules_display['antecedents'] = rules_display['antecedents'].apply(lambda x: ', '.join(x))
rules_display['consequents'] = rules_display['consequents'].apply(lambda x: ', '.join(x))

print(f'\nSo rules: {len(rules)}')
print(rules_display.round(3).to_string())

print('\nDoc ket qua:')
if len(rules) > 0:
    r = rules.iloc[0]
    ant = ', '.join(r['antecedents'])
    con = ', '.join(r['consequents'])
    print(f'  Rule: {ant} -> {con}')
    print(f'  Support: {r["support"]:.3f} ({r["support"]*100:.1f}% don hang co ca 2)')
    print(f'  Confidence: {r["confidence"]:.3f} ({r["confidence"]*100:.1f}% mua {ant} cung mua {con})')
    print(f'  Lift: {r["lift"]:.3f} ({"Lien quan!" if r["lift"] > 1 else "Doc lap"})')
````

## 3. Thực Hành: Phân Tích Giỏ Hàng Siêu Thị

````python
# Load du lieu
df_grocery = pd.read_csv('../Bai thi thu/Course Files/Groceries_dataset.csv')
print(f'Shape: {df_grocery.shape}')
print(f'Columns: {df_grocery.columns.tolist()}')
print(df_grocery.head(10))
````

````python
# EDA
print(f'So khach hang: {df_grocery["Member_number"].nunique()}')
print(f'So san pham: {df_grocery["itemDescription"].nunique()}')
print(f'\nTop 15 san pham ban chay:')

top_items = df_grocery['itemDescription'].value_counts().head(15)
fig, ax = plt.subplots(figsize=(10, 5))
top_items.plot(kind='barh', ax=ax, color='steelblue')
ax.set_xlabel('So luot mua')
ax.set_title('Top 15 san pham ban chay', fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# Tao basket: moi dong = 1 gio hang (group by Member + Date)
baskets = df_grocery.groupby(['Member_number', 'Date'])['itemDescription'].apply(list).reset_index()
print(f'So gio hang: {len(baskets)}')
print(f'Trung binh {baskets["itemDescription"].apply(len).mean():.1f} san pham/gio')

# One-hot encoding
te = TransactionEncoder()
te_data = te.fit_transform(baskets['itemDescription'])
df_basket = pd.DataFrame(te_data, columns=te.columns_)
print(f'\nBasket matrix: {df_basket.shape}')
````

````python
# Apriori
freq = apriori(df_basket, min_support=0.02, use_colnames=True)
print(f'Frequent itemsets: {len(freq)}')

# Rules
rules_grocery = association_rules(freq, metric='lift', min_threshold=1.0)
rules_grocery = rules_grocery.sort_values('lift', ascending=False)
print(f'Rules (lift > 1): {len(rules_grocery)}')

# Top rules
top_rules = rules_grocery.head(15).copy()
top_rules['antecedents'] = top_rules['antecedents'].apply(lambda x: ', '.join(x))
top_rules['consequents'] = top_rules['consequents'].apply(lambda x: ', '.join(x))
print('\nTop 15 rules theo Lift:')
print(top_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].round(3).to_string())
````

````python
# Visualization: Support vs Confidence vs Lift
if len(rules_grocery) > 0:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    sc1 = axes[0].scatter(rules_grocery['support'], rules_grocery['confidence'], 
                          c=rules_grocery['lift'], cmap='RdYlGn', s=40, alpha=0.7)
    axes[0].set_xlabel('Support'); axes[0].set_ylabel('Confidence')
    axes[0].set_title('Support vs Confidence\n(mau = Lift)', fontweight='bold')
    plt.colorbar(sc1, ax=axes[0], label='Lift')
    
    sc2 = axes[1].scatter(rules_grocery['support'], rules_grocery['lift'],
                          c=rules_grocery['confidence'], cmap='RdYlGn', s=40, alpha=0.7)
    axes[1].set_xlabel('Support'); axes[1].set_ylabel('Lift')
    axes[1].set_title('Support vs Lift\n(mau = Confidence)', fontweight='bold')
    axes[1].axhline(y=1, color='red', ls='--', alpha=0.5, label='Lift=1 (doc lap)')
    axes[1].legend()
    plt.colorbar(sc2, ax=axes[1], label='Confidence')
    
    plt.tight_layout()
    plt.show()
    
    print('\nDE XUAT KINH DOANH:')
    for _, r in rules_grocery.head(5).iterrows():
        ant = ', '.join(r['antecedents'])
        con = ', '.join(r['consequents'])
        print(f'  Mua [{ant}] -> de xuat [{con}] (Lift={r["lift"]:.2f}, Conf={r["confidence"]:.0%})')
````

## 4. Nguyên Tắc Chọn min_support và min_confidence

| Tham số | Thấp | Cao |
|:---|:---|:---|
| **min_support** | Nhiều itemsets nhưng rare | Ít itemsets, chỉ phổ biến |
| **min_confidence** | Nhiều rules nhưng yếu | Ít rules nhưng mạnh |

### Hướng dẫn thực tế:
- **min_support**: 0.01-0.05 (1-5% giao dịch)
- **min_confidence**: 0.3-0.7 (30-70%)
- **Lọc theo Lift > 1**: chỉ giữ rules THỰC SỰ liên quan
- Quan trọng: support CAO + confidence CAO + lift > 1 = **RULE TỐT**

---

## TONG HOP: 20% Kien Thuc -> 80% Ung Dung

| # | Kiến thức cốt lõi | Chi tiết |
|:--|:---|:---|
| 1 | **Support, Confidence, Lift** | 3 chỉ số PHẢI NHỚ. Lift > 1 = liên quan thật |
| 2 | **TransactionEncoder** | Chuyển list giao dịch → one-hot boolean |
| 3 | **apriori()** → **association_rules()** | 2 bước: tìm itemsets → sinh rules |
| 4 | **min_support ~ 0.01-0.05** | Tùy domain. Bắt đầu 0.02, điều chỉnh |
| 5 | **Lọc Lift > 1** luôn | Lift <= 1 = không có giá trị business |

### Code Template:
```python
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
df_onehot = pd.DataFrame(te.fit_transform(transactions), columns=te.columns_)

freq = apriori(df_onehot, min_support=0.02, use_colnames=True)
rules = association_rules(freq, metric='lift', min_threshold=1.0)
rules.sort_values('lift', ascending=False)
```
