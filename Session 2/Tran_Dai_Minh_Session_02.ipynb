```python
from sklearn.linear_model import LinearRegression
import pandas as pd 
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
df_housing = pd.DataFrame({
    'Total_rooms': [880, 7099, 1467, 1274, 1627, 919, 2535, 3104, 2555],
    'Median_income': [8.3252, 8.3014, 7.2574, 5.6431, 3.8462, 4.0368, 3.6591, 3.12, 2.0804],
    'Median_house_value': [452600, 358500, 352100, 341300, 342200, 269700, 299200, 241400, 226700]
})


X = df_housing.drop(columns = 'Median_house_value')
y = df_housing['Median_house_value']
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size= 0.4 , random_state= 42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

linear_model = LinearRegression()
linear_model.fit(X_train_scaled , y_train)
y_predict = linear_model.predict(X_test_scaled)


scores = cross_val_score(linear_model , X, y , cv = 3 , scoring= 'neg_mean_squared_error')
print(f"mean squared error of the linear regression model with cross validation  {-scores.mean():.2f}")

```

```python
from sklearn.neighbors import KNeighborsClassifier
df_movies = pd.DataFrame({
    'Movie_Name': [
        'The Shawshank Redemption',
        'The Godfather',
        'The Dark Knight',
        'The Godfather: Part II',
        '12 Angry Men',
        'The Lord of the Rings: The Return of the King',
        'Pulp Fiction',
        'The Shawshank Redemption'
    ],
    'IMDB_Ratings': [9.3, 9.2, 9, 9, 9, 8.9, 8.9, 9.3],
    'Meta_Score': [80, 100, 84, 90, 96, 94, 94, 80],
    'Genre': [
        'Drama',
        'Crime, Drama',
        'Action, Crime, Drama',
        'Crime, Drama',
        'Crime, Drama',
        'Action, Adventure, Drama',
        'Crime, Drama',
        'Drama'
    ]
})

knn = KNeighborsClassifier(n_neighbors= 5)
X = df_movies[['IMDB_Ratings', 'Meta_Score']]
y = df_movies['Genre']
knn.fit(X, y)

new_movive = [[8.7 , 85]]
predicted_genre = knn.predict(new_movive)
print(f"Predicted gerne of the new movie : {predicted_genre[0]}")
```
